## Erddruckbeiwerte

### Erdruhedruck


```python
K_0 = (1-np.sin(varphi_k))* (1+np.sin(beta_))/ np.cos(beta_)

Cell()
```

### Aktiver Erddruckbeiwert

```python
K_a = np.cos(varphi_k + alpha)**2 / (np.cos(alpha)**2 * np.cos(delta-alpha) * (1 + (np.sin(varphi_k+delta) * np.sin(varphi_k-beta_)/(np.cos(delta-alpha)*np.cos(alpha+beta_)))**0.5)**2)
K_ah = np.cos(delta) * K_a
Cell()
```

### Passiver Erddruckbeiwert

```python
K_p = np.cos(varphi_k - alpha)**2 / (np.cos(alpha)**2 * np.cos(delta-alpha) * (1 - (np.sin(varphi_k-delta) * np.sin(varphi_k+beta_)/(np.cos(delta-alpha)*np.cos(alpha+beta_)))**0.5)**2)

K_ph = np.cos(delta) * K_p
with open("K_ph.pkl", "wb") as f:
    pickle.dump(K_ph, f)
Cell()
```
