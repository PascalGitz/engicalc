### Berechnung

Knickeigenschaften

```python
l_cr = 3 *m 
l = 3*m 
d = 0.2*m
d__inf = 0.05*m
Cell(rows=4)
```


Exzentrizität infolge Imperfektion:

```python
alpha_i = min(max(0.01 / np.sqrt(l.m), 1/300), 1/200)
e_0d = max(alpha_i * l_cr /2, d / 30).to(mm)

Cell(precision=3)
```

Massgebende Auswirkungen

```python
M_1d = 0*kNm
N_d = 600*kN

Cell()
```

Exzentrizität 1. Ordnung

```python
e_1d = (M_1d / N_d).to(m)

Cell()
```

Exzentrizität infolge der Verformung des Druckglieds

Integrationskonstante

```python
c =np.pi**2

Cell()
```

Exzentrizität mit der Näherungslösung der Krümmung

Eigenschaften der Bewehrung

```python
f_sd = 390.*N/mm**2
E_s = 205000.*N/mm**2
Cell(rows = 2)
```

```python
chi_d = (2 * f_sd/ ( E_s * (d - d__inf))).to(1/km)
e_2d = (chi_d * l_cr**2 / c).to(mm)

Cell()
```

Exzentrizität

```python
e_d = e_0d + e_1d + e_2d

Cell()
```

Bemessungswert des Biegemoments

```python
M_d = (N_d * e_d).to(kNm)
Cell()
```
